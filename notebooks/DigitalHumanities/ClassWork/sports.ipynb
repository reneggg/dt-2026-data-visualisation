```python
#can be used to work with tabular data
import pandas as pd
df = pd.read_csv("competition.csv")
df.head() # Show the first few rows
df.describe() #Shows what calculations are available for numbers. Basic statistics

```

```python
import matplotlib.pyplot as plt # matplotlib helps create charts and visualizations
df.plot(kind="bar",x="Name",y="Years_of_practice", legend=False)
plt.title("Years of practice per Athlete")
plt.ylabel("Years of practice")
plt.show()
```

```python
import matplotlib.pyplot as plt # matplotlib helps create charts and visualizations
df.plot(kind="bar",x="Name",y="Age", legend=False)
plt.title("Age per Athlete")
plt.ylabel("Age")
plt.show()
```

```python
df.plot(kind="pie",y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%") #Has to be defined or rather imported once
plt.title("Age per Athlete")
```

```python

import matplotlib.pyplot as plt
plt.scatter(df["Age"], df["Years_of_practice"],s=df["Years_of_practice"]*60, alpha=0.67) #Alpha = change transparency, s=df and *60 for example change bubble size
plt.title("Age per Years of Practice")
plt.xlabel("Age")
plt.ylabel("Years of practice")
```

```python
import seaborn as sns #use this to create more clean and formatted plots
sns.boxplot(data=df, x="Age")
plt.title("Distribution of athlete age")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 5)) # Fixed figure size 
df[["Age", "Years_of_practice"]].boxplot() # Select columns
plt.title("Box plot for age and years of practice")
plt.show()
```

```python
avg_age=df.groupby("Sport")["Age"].mean() 
avg_age.plot(kind="bar")
plt.show()
```

```python
avg_yearsofpractice=df.groupby("Sport")["Years_of_practice"].mean()
avg_yearsofpractice.plot(kind="bar")
plt.show()
```
