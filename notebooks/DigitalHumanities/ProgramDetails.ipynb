```python
#Create a chart showing who puts in the most effort. This can be found from the number of hours.
```

```python
import pandas as pd
df = pd.read_csv("Competencies.csv")
df.head() # Show the first few rows
df.describe() #Shows what calculations are available for numbers. Basic statistics
import matplotlib.pyplot as plt # matplotlib helps create charts and visualizations
```

```python
avg_hours=df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="bar", y="Hours_Studied", legend=False)
plt.ylabel("Hours Studied")
plt.show()
```

```python
#Draw a bubble chart based on any variable. Also draw a box plot for hours and grades
```

```python
import matplotlib.pyplot as plt
plt.scatter(df["Hours_Studied"], df["Grade"], alpha=0.67) #Alpha = change transparency, s=df and *60 for example change bubble size
plt.title("Hours studied vs grades")
plt.xlabel("Hours studied")
plt.ylabel("Grade")
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 7)) 
df[["Hours_Studied", "Grade"]].boxplot() 
plt.title("Box plot for hours studied and grade")
plt.show()
```
