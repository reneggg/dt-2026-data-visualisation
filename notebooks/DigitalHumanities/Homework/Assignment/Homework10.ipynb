```python
import urllib.request
from wordcloud import WordCloud
import matplotlib.pyplot as plt

text = urllib.request.urlopen("https://raw.githubusercontent.com/reneggg/homeworkbusiness/refs/heads/main/Modena1937.txt").read().decode("utf-8")

# Using a standard list (.split() creates a list)
stopwords_list = urllib.request.urlopen("https://raw.githubusercontent.com/reneggg/homeworkbusiness/refs/heads/main/estonian-stopwords-lemmas.txt").read().decode("utf-8").lower().split()

def color_function(word, font_size, position, orientation, random_state=None, **additional_args):
    if word.lower() in stopwords_list:
        return "red"
    return "green"

wc = WordCloud(width=600, height=600, background_color="white", color_func=color_function).generate_from_text(text)
plt.imshow(wc)
plt.axis("off")
plt.show()
```

Due to some reason, the image generation we used in class did not work for me, so I simply imported matplotlib. Because of that I also turned off the axises.

For this word cloud I wanted to generate both the "normal" words & names together with stopwords. Stopwords are coloured red, whilst "normals" green. This is defined in lines 10-13. If the stopwords_list condition gets met (more about that in the second paragraph from now), it will return red, otherwise green. The Estonian stopword text is taken from *Kristel Uiboaed. Eesti keele stoppsõnad / Estonian stop words. 2018. DOI: 10.15155/RE-48*.

The reason behind this is because, as per my history lectures and seminars, the clergy had a tendency to write in a very apologetic, highly-praising manner. In Estonian you would use the word "nõtke" to describe this, which roughly translates to supple or gracious. You did not want to offend the person you directed your message towards. Hence, a very delicate style was frequently used.

To find this "style" within Wilhelm's charter, I found an Estonian stopwords list (cited above), which I then requested to be opened via url. There I used the `.lower()` to convert any word to lower case. This is necessary for `def color_function` (lines 10-13) to work - it standardizes any variation of a stopword. For example, it would default And, AND, and, etc. all to lower-case.

**Result:** Unsurprisingly, stopwords such as "ja", "et", "oma", and many more were far frequent than the normal words. Prominent are the words "kes" (who) and "igaüks" (everybody / every person), which in first sight could be deemed insignificant, but when taking to consideration the aforementioned supple style, it adds up. There exists a clear tendency to combine many smaller sentences into one. This longer sentence is then divided by many commas and stopwords. Not to mention that conjunctions such as "ja" and "et" are also very common. Thus, there exists a basis to believe that medieval legates and clergy had a distinct writing style.

If we got slightly mathematical here, then there were 25 commas amongst the 8 sentences. THis means ~4 commas per sentence [(25+8)/8]. Now imagine reading a text which is roughly 16-17 sentences. Each includes 2 or so commas. Should be far easier to perceive, right?

However, one cannot make clear consensuses based off this one charter alone. Frankly, through translation(s) and many other means of interpreting former texts, it could have lost its initial style. Perhaps the nature of Estonian as a language made the text clunkier than what it would have been in latin. As a result, I made another wordcloud, that is focused solely on the "normal" words.

When it comes to interpretation, I can make the following conclusions:
1) Auxilliary verbs such as "olema -> on" (to be) are logically used the most. Given how stretched the sentences are, the necessity for such words are increased.
2) The words "härra" (mister), "apostlik" (apostolic), "Kristus(e)" (Christ) are examples of words that show most and foremost how important both Christ and politeness were within a medieval society.
3) "Vabadus" (freedom) & "seadust rikkuda" (break the law) refer to how Estonia's society was built up. There existed a specific code of law that must not be broken. Thus, the guarantee for freedom depends not only on the words of God, i.e. the pope, but also by the secular. This is where the word "keiser" (emperor) comes to play.
4) Furthermore the word "annetada" (to donate) should be paired with "pidalitõbised" (person suffering from leprosy). This refers to the now non-existant St. John's Almshouse. Health and hygenie were therefore of importance. What better way to feel cared for than in a church institution? The aroma of incense, the tight building walls being held together with God's hands, not to mention seeing the Holiness via pictures. It sure was a sight to withhold.
5) "Eestimaa" (The land of Estonia) was recognised as an entity during this time period. This shows that the term for Estonia had probably been wide-spread and accepted by then. Beforehand Estonia was simply referred to as an ambiguous geographical location such as Aesti and Eastlanda. (Mägi 2018, pp. 145-146)
